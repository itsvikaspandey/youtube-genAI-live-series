## 1. Python If, Else & Elif Statements

### What is a conditional statement?
A **conditional statement** lets your program make decisions. Just like in real life you might think *"IF it is raining, THEN I will take an umbrella, ELSE I will not"* — Python code can do the same thing.

### The three keywords
| Keyword | Meaning |
|---|---|
| `if` | Runs a block of code **only if** a condition is `True` |
| `elif` | Short for "else if" — checks another condition **if the previous one was False** |
| `else` | Runs when **none** of the above conditions were `True` |

### Syntax rules
- Every `if`, `elif`, `else` line ends with a colon `:`
- The code inside each block must be **indented** (usually 4 spaces). Python uses indentation instead of `{ }` like other languages.


````python
age = 16
````

````python
print(age)
````

````python
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote yet.")
````

````python
# Example 2: if...else
temperature = 25

if temperature > 25:
    print("It's a hot day.")
else:
    print("It's not that hot today.")
````

````python
# Example 3: if...elif...else (checking multiple conditions in order)
marks = 58

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Your grade is: {grade}")
````

### Comparison & Logical Operators used in conditions
| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `a == b` |
| `!=` | not equal to | `a != b` |
| `>`, `<` | greater/less than | `a > b` |
| `>=`, `<=` | greater/less or equal | `a >= b` |
| `and` | both conditions true | `a > 0 and b > 0` |
| `or` | at least one condition true | `a > 0 or b > 0` |
| `not` | reverses a condition | `not a > 0` |


````python
# Example 4: Using logical operators
username = "admin"
password = "1235"

if username == "admin" and password == "1234":
    print("Login successful!")
else:
    print("Invalid credentials.")
````

## 2. Nested Conditionals

### What does "nested" mean?
A **nested conditional** is simply an `if` statement placed **inside** another `if` (or `elif`/`else`) statement. Think of it like a decision tree — you check one thing first, and only if that's true, you check something more specific.

**Real-life analogy:** *"IF it is raining, THEN check IF you have an umbrella. IF yes, go outside. IF no, stay home."*


````python
# Example 1: Basic nesting
is_raining = False
has_umbrella = False

if is_raining:
    if has_umbrella:
        print("Go outside, you're covered!")
    else:
        print("Better stay home, you'll get wet.")
else:
    print("No rain, enjoy your day!")
````

**Tip:** Deeply nested conditionals (nesting 4-5 levels deep) make code hard to read. When possible, combine conditions with `and`/`or` instead. Compare:


````python
# Less readable (deep nesting)
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("You can drive.")

# More readable (combined condition) — does the same thing
if age >= 18 and has_license:
    print("You can drive.")
````

## 3. For Loops & While Loops

### Why do we need loops?
Loops let you **repeat** a block of code multiple times without writing it over and over. There are two main types in Python:

- **`for` loop** — used when you know how many times to repeat, or you want to go through each item in a collection (like a list or a range of numbers).
- **`while` loop** — used when you want to repeat something **until a condition becomes False** (you may not know exactly how many times in advance).


````python
# Example 1: for loop with range()
# range(5) generates numbers 0, 1, 2, 3, 4
for i in range(5):
    print("Iteration number:", i)
````

````python
# Example 2: for loop over a range with start, stop, step
for i in range(2, 11, 2):   # start=2, stop=11 (exclusive), step=2
    print(i)
````

````python
# Example 3: for loop over a string (loops go through characters too!)
for letter in "Python":
    print(letter)
````

````python
# Example 4: while loop — repeats until condition becomes False
count = 1

while count <= 5:
    print("Count is:", count)
    count = count + 1   # IMPORTANT: always update the variable, or you'll get an infinite loop!
````

### `for` vs `while` — when to use which?
| Situation | Use |
|---|---|
| You know the exact number of repetitions | `for` |
| You're iterating over a list, string, or range | `for` |
| You need to repeat "until something happens" (unknown count) | `while` |
| Example: keep asking for input until user enters a valid one | `while` |

### ⚠️ Common beginner mistake: Infinite loops
A `while` loop **must** eventually make its condition `False`, otherwise it runs forever and freezes your program.


````python
# BAD (do NOT run this — infinite loop, shown only as example, kept commented)
# count = 1
# while count <= 5:
#     print(count)     # count never changes, so condition is always True forever

# GOOD — count is updated every time, so loop eventually stops
count = 1
while count <= 5:
    print(count)
    count += 1
````

## 4. Break, Continue & Pass

These three keywords give you extra control **inside** loops (and `pass` works inside conditionals/functions too).

| Keyword | What it does |
|---|---|
| `break` | Immediately **exits** the loop entirely |
| `continue` | **Skips** the rest of the current iteration and moves to the next one |
| `pass` | Does **nothing** — it's a placeholder so your code doesn't error out on an empty block |


````python
# Example 1: break — stop the loop as soon as we find what we need
numbers = [4, 8, 15, 16, 23, 42]

for num in numbers:
    if num == 16:
        print("Found 16! Stopping the search.")
        break
    print("Checking:", num)
````

````python
# Example 2: continue — skip even numbers, only print odd numbers
for num in range(1, 11):
    if num % 2 == 0:
        continue   # skip the rest of this iteration
    print(num)
````

````python
# Example 3: pass — used as a placeholder when code is not written yet
for num in range(5):
    if num == 3:
        pass   # TODO: handle this case later, but don't crash for now
    else:
        print(num)

# pass is also handy in empty function/class definitions during planning:
def my_future_function():
    pass   # will implement this later
````

## 5. Python Lists & Common List Operations

### What is a list?
A **list** is an ordered collection that can hold multiple values (numbers, strings, even other lists) in a single variable. Lists are written with **square brackets `[ ]`** and items separated by commas.

```python
fruits = ["apple", "banana", "cherry"]
```

Lists are:
- **Ordered** — items keep their position (index).
- **Mutable** — you can change, add, or remove items after creation.
- **Zero-indexed** — the first item is at position `0`, not `1`.


````python
# Example 1: creating and accessing a list
fruits = ["apple", "banana", "cherry", "date"]

print(fruits)          # the whole list
print(fruits[0])       # first item -> apple
print(fruits[-1])      # last item -> date (negative index counts from the end)
print(fruits[1:3])     # slicing: items from index 1 up to (not including) 3
````

````python
# Example 2: modifying a list
fruits = ["apple", "banana", "cherry"]

fruits[1] = "blueberry"     # replace an item
print(fruits)

fruits.append("date")       # add an item to the end
print(fruits)

fruits.insert(1, "mango")   # insert at a specific position
print(fruits)

fruits.remove("apple")      # remove a specific item by value
print(fruits)

removed_item = fruits.pop()  # remove and return the LAST item
print("Removed:", removed_item)
print(fruits)
````

### Common List Methods & Functions
| Method / Function | What it does | Example |
|---|---|---|
| `.append(x)` | adds `x` to the end | `lst.append(5)` |
| `.insert(i, x)` | inserts `x` at index `i` | `lst.insert(0, 5)` |
| `.remove(x)` | removes the first occurrence of `x` | `lst.remove(5)` |
| `.pop(i)` | removes & returns item at index `i` (default: last) | `lst.pop()` |
| `.sort()` | sorts the list in place (ascending by default) | `lst.sort()` |
| `.reverse()` | reverses the list in place | `lst.reverse()` |
| `len(lst)` | number of items in the list | `len(lst)` |
| `sum(lst)` | sum of all numeric items | `sum(lst)` |
| `max(lst)` / `min(lst)` | largest / smallest item | `max(lst)` |
| `x in lst` | checks if `x` exists in the list | `5 in lst` |
| `.count(x)` | counts occurrences of `x` | `lst.count(5)` |
| `.index(x)` | finds the index of the first `x` | `lst.index(5)` |


````python
# Example 3: common list operations in action
numbers = [4, 2, 9, 1, 5, 9]

print("Length:", len(numbers))
print("Sum:", sum(numbers))
print("Max:", max(numbers))
print("Min:", min(numbers))
print("Does 9 exist?", 9 in numbers)
print("How many 9s?", numbers.count(9))
print("Index of first 9:", numbers.index(9))

numbers.sort()
print("Sorted:", numbers)

numbers.reverse()
print("Reversed:", numbers)
````

````python
# Example 4: combining two lists and creating a new one
list_a = [1, 2, 3]
list_b = [4, 5, 6]

combined = list_a + list_b     # concatenation
print("Combined:", combined)

repeated = list_a * 3          # repetition
print("Repeated:", repeated)
````

## 6. List Iteration using Loops

Since lists are ordered collections, `for` loops are the most natural way to **go through (iterate)** every item one by one.


````python
# Example 1: basic iteration
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("I like", fruit)
````

````python
# Example 2: iterating with index using enumerate()
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
````

````python
# Example 3: iterating with a while loop (less common, but good to know)
fruits = ["apple", "banana", "cherry"]
i = 0

while i < len(fruits):
    print(fruits[i])
    i += 1
````

````python
# Example 4: building a new list while iterating (very common pattern)
numbers = [1, 2, 3, 4, 5]
squares = []

for n in numbers:
    squares.append(n ** 2)

print(squares)

# The same result can be written as a "list comprehension" (a compact one-liner)
squares_comprehension = [n ** 2 for n in numbers]
print(squares_comprehension)
````

````python
# Example 5: iterating with conditionals inside the loop
numbers = [10, 15, 22, 33, 40, 51]
even_numbers = []

for n in numbers:
    if n % 2 == 0:
        even_numbers.append(n)

print("Even numbers:", even_numbers)
````

### Exercise 1 — FizzBuzz (Difficulty: Easy)
**Question:** Print numbers from 1 to 20. But for multiples of 3, print `"Fizz"` instead of the number, for multiples of 5 print `"Buzz"`, and for multiples of both 3 and 5 print `"FizzBuzz"`.

*Why interviewers ask this:* It tests whether you understand `if/elif/else`, loops, and the modulo operator (`%`) — one of the most common warm-up interview questions.


````python
# --- Try it yourself first! ---
````

**Solution & Explanation:**
- We loop through numbers 1 to 20 using `range(1, 21)`.
- We check the **most specific condition first** (divisible by both 3 and 5), because if we checked "divisible by 3" first, `15` would print `"Fizz"` instead of `"FizzBuzz"`.
- `%` (modulo) gives the remainder of a division. `n % 3 == 0` means `n` is exactly divisible by 3.


````python
for n in range(1, 21):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
````

### Exercise 2 — Find the Second Largest Number in a List (Difficulty: Medium)
**Question:** Given a list of numbers, find the second largest number **without using the built-in `sorted()` function**.

*Why interviewers ask this:* Tests loop logic, tracking multiple variables, and edge-case handling (duplicates).


````python
# --- Try it yourself first! ---
numbers = [12, 45, 2, 41, 45, 100, 99]
````

**Solution & Explanation:**
- We keep track of two variables: `largest` and `second_largest`.
- As we loop through the list, if we find a number bigger than `largest`, the old `largest` becomes the new `second_largest`, and the number becomes the new `largest`.
- If a number is smaller than `largest` but bigger than `second_largest`, it only updates `second_largest`.
- We start both trackers at negative infinity so any real number will be larger on the first comparison.


````python
numbers = [12, 45, 2, 41, 45, 100, 99]

largest = float('-inf')
second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print("Largest:", largest)
print("Second Largest:", second_largest)
````

### Exercise 3 — Check if a Number is Prime (Difficulty: Medium)
**Question:** Write code that checks whether a given number is a **prime number** (a number greater than 1 that has no divisors other than 1 and itself).

*Why interviewers ask this:* Combines loops, conditionals, and `break` for early exit — very common fundamentals question.


````python
# --- Try it yourself first! ---
number = 29
````

**Solution & Explanation:**
- Numbers 0 and 1 are not prime by definition.
- We only need to check divisors from 2 up to the number itself (in practice, up to its square root is more efficient, but a simple loop is fine for beginners).
- We use `break` to stop checking as soon as we find any divisor — no need to keep going.
- We use a flag variable `is_prime` to track the result.


````python
number = 29
is_prime = True

if number < 2:
    is_prime = False
else:
    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break   # no need to check further once we find a divisor

if is_prime:
    print(f"{number} is a prime number.")
else:
    print(f"{number} is NOT a prime number.")
````

### Exercise 4 — Remove Duplicates from a List while Preserving Order (Difficulty: Medium)
**Question:** Given a list with duplicate values, return a new list with duplicates removed, **keeping the original order** of first appearance.

*Why interviewers ask this:* A very common real-world data-cleaning task; tests list iteration + membership checking (`in`).


````python
# --- Try it yourself first! ---
items = [1, 3, 2, 3, 5, 1, 4, 2]
````

**Solution & Explanation:**
- We create an empty list `unique_items` to store results.
- We loop through the original list, and for each item we check `if item not in unique_items` — only adding it if it hasn't been seen yet.
- Using `set(items)` alone would remove duplicates but **not preserve order**, which is why we do it manually here.


````python
items = [1, 3, 2, 3, 5, 1, 4, 2]
unique_items = []

for item in items:
    if item not in unique_items:
        unique_items.append(item)

print(unique_items)
````

### Exercise 5 — Reverse a List Without Using `.reverse()` or Slicing (Difficulty: Hard)
**Question:** Reverse a list manually, without using `lst.reverse()`, `lst[::-1]`, or `reversed()`.

*Why interviewers ask this:* Tests understanding of loop indices and swapping — a classic algorithmic thinking exercise.


````python
# --- Try it yourself first! ---
data = [10, 20, 30, 40, 50]
````

**Solution & Explanation:**
- We use a `while` loop with two pointers: `left` starting at the beginning (index 0) and `right` starting at the end (`len(data) - 1`).
- On each iteration, we **swap** the elements at `left` and `right`, then move `left` forward and `right` backward.
- The loop stops when `left` and `right` meet or cross — at that point the whole list has been reversed in place.


````python
data = [10, 20, 30, 40, 50]

left = 0
right = len(data) - 1

while left < right:
    data[left], data[right] = data[right], data[left]   # swap
    left += 1
    right -= 1

print(data)
````
