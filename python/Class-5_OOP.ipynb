## 1. Why Object-Oriented Programming Matters for GenAISo far, every program we've written has been a list of instructions running top to bottom — using functions to organize logic. That works fine for small tasks, but real applications need to **track state** (data that changes and needs to be remembered) alongside the logic that acts on it.**Real-world analogy:** think of a class as a *cookie cutter* and each object as a *cookie*. The cutter (class) defines the shape — every cookie made from it has the same shape, but each individual cookie (object) is separate and can be decorated differently.Why this matters for GenAI specifically — every major AI library you'll use later in this series is built almost entirely out of classes:- An LLM client is an object (`client = OpenAI()`)- A RAG vector store is an object- An AI agent, its memory, and its tools are all objects talking to each otherA chatbot needs to *remember* conversation history, *track* its settings, and *expose* actions like `.send_message()` — all bundled together. That bundling of data + behavior is exactly what a class gives you.

## 2. Classes and Objects- **Class** = the blueprint / template (the cookie cutter)- **Object** (or *instance*) = a specific thing built from that blueprint (an actual cookie)You can create as many objects as you want from one class, and each one is completely independent of the others.

```python
class Chatbot:    pass  # an empty class for now — just the blueprint, nothing inside yet# Creating (instantiating) two separate objects from the same classbot1 = Chatbot()bot2 = Chatbot()print(type(bot1))       # tells us bot1 is a Chatbot objectprint(bot1 is bot2)     # False — two completely separate objects, even though same class
```

## 3. `__init__`, `self`, Instance Attributes and Methods**`__init__`** is a special method (Python calls these "dunder" methods, short for double-underscore) that runs **automatically** the moment an object is created. It's where you set up the object's starting data.**`self`** refers to "this specific object." Every method inside a class takes `self` as its first parameter so Python knows *which* object's data to work with — it's automatically passed in when you call a method, you never type it yourself when calling.Think of it like filling out a form: `__init__` is the form, and `self` is "this particular form, not anyone else's."

```python
class Chatbot:    def __init__(self, name, model="gpt-3.5"):        # Everything below is stored ON this specific object using self.        self.name = name          # instance attribute — this bot's name        self.model = model        # instance attribute — this bot's model        self.history = []         # instance attribute — starts empty for every new bot    def send_message(self, message):   # instance method — self gives access to THIS bot's data        self.history.append(message)        return f"[{self.name}] Received: {message}"# When you write Chatbot("Blueprint Assistant"), Python automatically:# 1. Creates a new empty object# 2. Calls __init__(self=that_new_object, name="Blueprint Assistant")bot = Chatbot("Blueprint Assistant")print(bot.name)                          # accessing an instance attributeprint(bot.send_message("Explain recursion"))  # calling an instance methodprint(bot.history)                       # history remembers what was sent
```

```python
# Proof that self keeps each object's data separatebot_a = Chatbot("Bot A")bot_b = Chatbot("Bot B")bot_a.send_message("Hello from A")bot_b.send_message("Hello from B")print("Bot A's history:", bot_a.history)  # only has A's messageprint("Bot B's history:", bot_b.history)  # only has B's message — completely separate
```

## 4. Class vs Instance Variables- **Instance variable** — belongs to ONE object only (set using `self.something = ...` inside `__init__`, like `name` above)- **Class variable** — shared by EVERY object of that class, defined directly inside the class body (not inside `__init__`)Use a class variable when the value should be the *same for every object* — like a running counter of how many bots exist.

```python
class Chatbot:    total_bots_created = 0  # CLASS variable — one shared copy for ALL Chatbot objects    def __init__(self, name):        self.name = name              # INSTANCE variable — unique to each object        Chatbot.total_bots_created += 1   # update the shared class variablebot1 = Chatbot("Bot A")bot2 = Chatbot("Bot B")bot3 = Chatbot("Bot C")print("Total bots created:", Chatbot.total_bots_created)  # 3 — shared across all objectsprint(bot1.name, bot2.name, bot3.name)                     # each has its OWN name
```

## 5. Inheritance and Method OverridingInheritance lets one class (the **child**) reuse everything from another class (the **parent**), and add or change behavior on top. This avoids rewriting the same code twice.**Real-world analogy:** a `SportsCar` is still a `Car` — it has everything a car has (wheels, engine, `drive()`), plus extra stuff (a turbo boost). You wouldn't rewrite "car" from scratch just to describe a sports car.`super().__init__(...)` means "run the parent class's `__init__` first, then add anything extra." **Method overriding** means the child class defines a method with the *same name* as the parent, replacing that specific behavior.

```python
class Chatbot:                          # PARENT class    def __init__(self, name):        self.name = name    def send_message(self, message):        return f"[{self.name}] Basic reply to: {message}"class RAGChatbot(Chatbot):              # CHILD class — inherits from Chatbot    def __init__(self, name, knowledge_base):        super().__init__(name)          # reuse the parent's __init__ (sets self.name)        self.knowledge_base = knowledge_base   # then add something new    def send_message(self, message):    # OVERRIDING — same method name, new behavior        return f"[{self.name}] Searching '{self.knowledge_base}' for: {message}"basic_bot = Chatbot("Basic Bot")rag_bot = RAGChatbot("RAG Bot", "company_docs")print(basic_bot.send_message("What is Python?"))   # uses Chatbot's versionprint(rag_bot.send_message("What is Python?"))      # uses RAGChatbot's overridden version# RAGChatbot still has everything Chatbot has, for free:print(rag_bot.name)
```

## 6. Encapsulation — Public, Protected & PrivateEncapsulation means controlling how a class's internal data can be accessed or changed from outside, instead of leaving everything wide open. Python signals this through naming convention (it doesn't strictly enforce it, unlike some other languages):| Naming | Meaning | Example ||---|---|---|| `name` | **Public** — anyone can access/change it freely | `self.name` || `_name` | **Protected** — a signal of "internal use, please don't touch this directly" (convention only) | `self._temperature` || `__name` | **Private** — Python actively makes this harder to access from outside the class (name-mangled) | `self.__api_key` |`@property` lets you add validation logic while still letting the outside code use it like a normal attribute (no parentheses needed).

```python
class Chatbot:    def __init__(self, name, temperature=0.7):        self.name = name        self._temperature = temperature   # protected — internal, but add validation via property below    @property    def temperature(self):        return self._temperature    @temperature.setter    def temperature(self, value):        # This validation runs EVERY time someone tries to set bot.temperature        if not (0 <= value <= 1):            raise ValueError("Temperature must be between 0 and 1")        self._temperature = valuebot = Chatbot("Bot A")print(bot.temperature)   # 0.7 — reads through the @property, looks like a normal attributebot.temperature = 0.9    # goes through the setter, passes validationprint(bot.temperature)bot.temperature = 5      # invalid — the setter blocks this and raises an error
```

## 7. PolymorphismPolymorphism means **different classes can respond to the same method call in their own way**. You already saw a hint of this with method overriding — polymorphism is that same idea, used across a group of related objects.**Real-world analogy:** you can tell many different animals to `.make_sound()` — a dog barks, a cat meows — same instruction, different behavior depending on which object you're talking to. You don't need to know exactly which animal it is to call the method; you just trust each one knows how to respond correctly.

```python
class TextChatbot(Chatbot):    def respond(self):        return f"{self.name} (text): Here's a written answer."class VoiceChatbot(Chatbot):    def respond(self):        return f"{self.name} (voice): Here's a spoken answer."bots = [    TextChatbot("Assistant A"),    VoiceChatbot("Assistant B"),]# Same method call — .respond() — different behavior depending on the actual object.# This is polymorphism: we loop through without caring which exact class each bot is.for bot in bots:    print(bot.respond())
```

```python
# Solutionclass BankAccount:    def __init__(self, owner, balance=0):        self.owner = owner        self.balance = balance    def deposit(self, amount):        self.balance += amount        return self.balance    def withdraw(self, amount):        if amount > self.balance:            raise ValueError("Insufficient balance")        self.balance -= amount        return self.balanceacc = BankAccount("Vikas", 1000)acc.deposit(500)print(acc.balance)try:    acc.withdraw(5000)except ValueError as e:    print("Error:", e)
```

### Exercise 2 — Polymorphism with Shapes (Difficulty: Medium)Create a base class `Shape` with a method `area()` that returns 0. Create two child classes, `Square` (takes `side`) and `Circle` (takes `radius`), each overriding `area()` with the correct formula. Loop through a list of both shape types and print each one's area — without checking which class it is.

```python
# Solutionimport mathclass Shape:    def area(self):        return 0class Square(Shape):    def __init__(self, side):        self.side = side    def area(self):        return self.side ** 2class Circle(Shape):    def __init__(self, radius):        self.radius = radius    def area(self):        return round(math.pi * self.radius ** 2, 2)shapes = [Square(4), Circle(3)]for shape in shapes:    # polymorphism: same .area() call, correct formula runs automatically    print(f"{type(shape).__name__} area: {shape.area()}")
```
