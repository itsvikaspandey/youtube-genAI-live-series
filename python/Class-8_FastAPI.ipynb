## 1. What is an API?

An **API (Application Programming Interface)** allows two software systems to communicate.

A typical web API flow is:

```text
Client
   │
   │ HTTP Request
   ▼
FastAPI Application
   │
   │ Python Logic
   ▼
Data / Database / AI Model
   │
   │ Response
   ▼
Client
```

### Example

A client sends:

```http
GET /users/101
```

The server processes the request and returns:

```json
{
    "id": 101,
    "name": "Vik"
}
```

### Why APIs matter in GenAI

Your frontend, mobile app, another service, or an AI workflow can communicate with your Python AI application through an API.


## 2. HTTP Methods
### REST basics
REST is a style of designing web APIs around resources. For example, a Notes API treats notes as resources. The URL identifies a resource and the HTTP method describes the requested operation.
The most common HTTP methods we'll use:

| Method | Typical purpose |
|---|---|
| GET | Read data |
| POST | Create data / send data for processing |
| PUT | Replace/update a resource |
| PATCH | Partially update a resource |
| DELETE | Delete a resource |

For our Notes API:

```text
GET    /notes
POST   /notes
GET    /notes/{note_id}
PUT    /notes/{note_id}
DELETE /notes/{note_id}
```

## 8. Path Parameters

A path parameter is part of the URL.

Example:

```text
GET /notes/10
```

Here, `10` is the `note_id`.


````python
from fastapi import FastAPI

app = FastAPI()


@app.get("/notes/{note_id}")
def get_note(note_id: int):
    return {
        "id": note_id,
        "title": f"Note {note_id}"
    }
````

### Type validation

Because we wrote:

```python
note_id: int
```

FastAPI knows that `note_id` should be an integer.

Try:

```text
/notes/10
```

and then:

```text
/notes/abc
```

FastAPI automatically generates a validation error for the second request.


## 9. Query Parameters

Query parameters appear after `?`.

Example:

```text
GET /notes?limit=5
```

or:

```text
GET /notes?limit=5&search=python
```


````python
from fastapi import FastAPI

app = FastAPI()


@app.get("/notes")
def get_notes(limit: int = 10, search: str | None = None):
    return {
        "limit": limit,
        "search": search
    }
````

### Path vs Query parameter

```text
/notes/10
       ↑
   Path parameter


/notes?limit=10
       ↑
   Query parameter
```

**Rule of thumb:**

- Path → identifies a specific resource
- Query → filters, searches, sorts, or controls the response


## 10. Pydantic & Request Body

For POST requests, we often send JSON data.

Pydantic models define the expected structure.

```text
Client
   │
   │ JSON
   ▼
FastAPI
   │
   │ Validation
   ▼
Pydantic Model
   │
   ▼
Python Function
```


````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()


class Note(BaseModel):
    title: str
    content: str
    category: str = "general"


@app.post("/notes")
def create_note(note: Note):
    return {
        "message": "Note created",
        "note": note
    }
````

### Example request

```json
{
    "title": "Learn FastAPI",
    "content": "Build APIs using Python",
    "category": "python"
}
```

FastAPI + Pydantic will validate the incoming data before your function runs.


## 11. Required vs Optional Fields

Pydantic lets us clearly describe our data.



````python
from pydantic import BaseModel
from typing import Optional


class User(BaseModel):
    name: str
    age: int
    email: Optional[str] = None
````

Here:

- `name` → required
- `age` → required
- `email` → optional
- `email` defaults to `None`

This type of validation becomes extremely useful when building AI APIs.


## 12. PUT — Update a Resource

PUT is commonly used to update or replace an existing resource.


````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()


class Note(BaseModel):
    title: str
    content: str


@app.put("/notes/{note_id}")
def update_note(note_id: int, note: Note):
    return {
        "id": note_id,
        "message": "Note updated",
        "note": note
    }
````

## 13. PATCH — Partial Update

PATCH is useful when only some fields need to change.

Conceptually:

```text
PUT
→ Replace/update the resource

PATCH
→ Change only selected fields
```

For the first FastAPI class, focus on understanding the difference rather than building a complex PATCH implementation.


## 14. DELETE

Let's create a delete endpoint.

## 15. Response Models

Request models define what the client **sends**.

Response models define what the API **returns**.



````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()


class UserResponse(BaseModel):
    id: int
    name: str


@app.get("/user", response_model=UserResponse)
def get_user():
    return {
        "id": 101,
        "name": "Vik",
        "internal_note": "This should not be exposed"
    }
````

### Why response models matter

They help you:

- Keep API responses consistent
- Validate output
- Control what data is exposed
- Generate better documentation


## 16. Error Handling with HTTPException

You already learned Python error handling.

Now let's see how errors work in an API.


````python
from fastapi import FastAPI, HTTPException

app = FastAPI()


notes = {
    1: {"id": 1, "title": "FastAPI"},
    2: {"id": 2, "title": "GenAI"},
}


@app.get("/notes/{note_id}")
def get_note(note_id: int):
    if note_id not in notes:
        raise HTTPException(
            status_code=404,
            detail="Note not found"
        )

    return notes[note_id]
````

### Important idea

Don't return:

```text
"Something went wrong"
```

for every failure.

Use meaningful HTTP status codes and error messages so clients know what happened.


## 17. Building Our Mini Project — Notes API

Now let's combine the concepts.

### Requirements

Our API should support:

```text
POST   /notes
GET    /notes
GET    /notes/{note_id}
PUT    /notes/{note_id}
DELETE /notes/{note_id}
```

We'll initially store data in memory.

```text
Python dictionary
       ↓
FastAPI
       ↓
API endpoints
```

No database yet — we will add that in a later class.


````python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI(title="Notes API")


class Note(BaseModel):
    title: str
    content: str
    category: str = "general"


notes = {}
next_id = 1


@app.post("/notes", status_code=201)
def create_note(note: Note):
    global next_id

    notes[next_id] = {
        "id": next_id,
        **note.model_dump()
    }

    created_note = notes[next_id]
    next_id += 1

    return created_note


@app.get("/notes")
def get_notes():
    return list(notes.values())


@app.get("/notes/{note_id}")
def get_note(note_id: int):
    if note_id not in notes:
        raise HTTPException(
            status_code=404,
            detail="Note not found"
        )

    return notes[note_id]


@app.put("/notes/{note_id}")
def update_note(note_id: int, note: Note):
    if note_id not in notes:
        raise HTTPException(
            status_code=404,
            detail="Note not found"
        )

    notes[note_id] = {
        "id": note_id,
        **note.model_dump()
    }

    return notes[note_id]


@app.delete("/notes/{note_id}")
def delete_note(note_id: int):
    if note_id not in notes:
        raise HTTPException(
            status_code=404,
            detail="Note not found"
        )

    deleted_note = notes.pop(note_id)

    return {
        "message": "Note deleted",
        "note": deleted_note
    }
````

## 19. Routers — Organizing a Real Project

As applications grow, keeping every endpoint inside `main.py` becomes difficult.

FastAPI provides `APIRouter`.

A simple structure:

```text
app/
├── main.py
├── routers/
│   └── notes.py
├── schemas/
│   └── note.py
└── services/
    └── note_service.py
```

The idea is:

```text
main.py
   ↓
Router
   ↓
Service / Business Logic
   ↓
Data / Database / AI Model
```


````python
# routers/notes.py

from fastapi import APIRouter

router = APIRouter(prefix="/notes", tags=["Notes"])


@router.get("/")
def get_notes():
    return {"message": "All notes"}
````

````python
# main.py

from fastapi import FastAPI
from routers.notes import router as notes_router

app = FastAPI()

app.include_router(notes_router)
````

## 20. Dependencies — Introduction

FastAPI has a powerful dependency injection system.

For example:

```python
from fastapi import Depends
```

A dependency can provide reusable functionality such as:

- Authentication
- Database sessions
- Common parameters
- Shared validation
- Services

For this class, understand the **idea** of `Depends()`.

We'll explore dependency injection in more depth when we build larger AI applications.


````python
from fastapi import Depends, FastAPI

app = FastAPI()


def get_common_message():
    return "Reusable dependency"


@app.get("/demo")
def demo(message: str = Depends(get_common_message)):
    return {"message": message}
````
