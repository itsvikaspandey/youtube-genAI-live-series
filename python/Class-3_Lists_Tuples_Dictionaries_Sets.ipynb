## 1. Python Lists & Common List Operations

````python
# Example 3: common list operations in action
numbers = [4, 2, 9, 1, 5, 9]

print("Length:", len(numbers))
print("Sum:", sum(numbers))
print("Max:", max(numbers))
print("Min:", min(numbers))
print("Does 9 exist?", 9 in numbers)
print("How many 9s?", numbers.count(9))
print("Index of first 9:", numbers.index(9))

numbers.sort()
print("Sorted:", numbers)

numbers.reverse()
print("Reversed:", numbers)
````

````python
# Example 4: combining two lists and creating a new one
list_a = [1, 2, 3]
list_b = [4, 5, 6]

combined = list_a + list_b     # concatenation
print("Combined:", combined)

repeated = list_a * 3          # repetition
print("Repeated:", repeated)
````

## 2. List Iteration using Loops

Since lists are ordered collections, `for` loops are the most natural way to **go through (iterate)** every item one by one.


````python
# Example 1: basic iteration
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

````python
# Example 2: iterating with index using enumerate()
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
````

````python
# Example 3: iterating with a while loop (less common, but good to know)
fruits = ["apple", "banana", "cherry"]
i = 0

while i < len(fruits):
    print(fruits[i])
    # i += 1 
    i = i + 1
````

````python
# Example 4: building a new list while iterating (very common pattern)
numbers = [1, 2, 3, 4, 5]
squares = []

for n in numbers:
    squares.append(n ** 2)

print(squares)

# The same result can be written as a "list comprehension" (a compact one-liner)
squares_comprehension = [n ** 2 for n in numbers]
print(squares_comprehension)
````

````python
# Example 5: iterating with conditionals inside the loop
numbers = [10, 15, 22, 33, 40, 51]
even_numbers = []

for n in numbers:
    if n % 2 == 0:
        even_numbers.append(n)

print("Even numbers:", even_numbers)
````

### Exercise 2 — Find the Second Largest Number in a List (Difficulty: Medium)
**Question:** Given a list of numbers, find the second largest number **without using the built-in `sorted()` function**.

*Why interviewers ask this:* Tests loop logic, tracking multiple variables, and edge-case handling (duplicates).


````python
# --- Try it yourself first! ---
numbers = [12, 45, 2, 41, 45, 100, 99]
````

**Solution & Explanation:**
- We keep track of two variables: `largest` and `second_largest`.
- As we loop through the list, if we find a number bigger than `largest`, the old `largest` becomes the new `second_largest`, and the number becomes the new `largest`.
- If a number is smaller than `largest` but bigger than `second_largest`, it only updates `second_largest`.
- We start both trackers at negative infinity so any real number will be larger on the first comparison.


````python
numbers = [12, 45, 2, 41, 45, 100, 99]

largest = float('-inf')
second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print("Largest:", largest)
print("Second Largest:", second_largest)
````

### Exercise 4 — Remove Duplicates from a List while Preserving Order (Difficulty: Medium)
**Question:** Given a list with duplicate values, return a new list with duplicates removed, **keeping the original order** of first appearance.

*Why interviewers ask this:* A very common real-world data-cleaning task; tests list iteration + membership checking (`in`).


````python
# --- Try it yourself first! ---
items = [1, 3, 2, 3, 5, 1, 4, 2]
````

**Solution & Explanation:**
- We create an empty list `unique_items` to store results.
- We loop through the original list, and for each item we check `if item not in unique_items` — only adding it if it hasn't been seen yet.
- Using `set(items)` alone would remove duplicates but **not preserve order**, which is why we do it manually here.


````python
items = [1, 3, 2, 3, 5, 1, 4, 2]
unique_items = []

for item in items:
    if item not in unique_items:
        unique_items.append(item)

print(unique_items)
````

### Exercise 5 — Reverse a List Without Using `.reverse()` or Slicing (Difficulty: Hard)
**Question:** Reverse a list manually, without using `lst.reverse()`, `lst[::-1]`, or `reversed()`.

*Why interviewers ask this:* Tests understanding of loop indices and swapping — a classic algorithmic thinking exercise.


````python
# --- Try it yourself first! ---
data = [10, 20, 30, 40, 50]
````

**Solution & Explanation:**
- We use a `while` loop with two pointers: `left` starting at the beginning (index 0) and `right` starting at the end (`len(data) - 1`).
- On each iteration, we **swap** the elements at `left` and `right`, then move `left` forward and `right` backward.
- The loop stops when `left` and `right` meet or cross — at that point the whole list has been reversed in place.


````python
data = [10, 20, 30, 40, 50]

left = 0
right = len(data) - 1

while left < right:
    data[left], data[right] = data[right], data[left]   # swap
    left += 1
    right -= 1

print(data)
````

## 3. Python Tuples & Common Tuple Operations

### What is a tuple?
A **tuple** is an ordered collection of items, just like a list — but with one key difference: **tuples are immutable**, meaning once created, you cannot change, add, or remove items.

Tuples are written with **parentheses `( )`**:
```python
point = (10, 20)
```

### Why use a tuple instead of a list?
- You want to guarantee the data **cannot be accidentally changed** later in your program (e.g., coordinates, RGB colors, days of the week).
- Tuples are slightly **faster** and use less memory than lists.
- Tuples can be used as **dictionary keys** (lists cannot, because lists are mutable).


````python
# Example 1: creating and accessing a tuple
coordinates = (10, 20, 30)

print(coordinates)
print(coordinates[0])      # first item -> 10
print(coordinates[-1])     # last item -> 30
print(coordinates[0:2])    # slicing works just like lists -> (10, 20)
````

````python
# Example 2: the "one-element tuple" trap
not_a_tuple = (5)         # this is just an integer! parentheses alone don't make a tuple
actual_tuple = (5,)       # the comma is what makes it a tuple

print(type(not_a_tuple))
print(type(actual_tuple))
````

````python
# Example 3: tuples are immutable - this will raise an error
point = (10, 20)

try:
    point[0] = 99
except TypeError as e:
    print("Error:", e)
````

### Common Tuple Operations
| Operation | What it does | Example |
|---|---|---|
| `len(t)` | number of items | `len((1,2,3))` -> `3` |
| `t.count(x)` | counts occurrences of `x` | `(1,2,2).count(2)` -> `2` |
| `t.index(x)` | finds index of first `x` | `(1,2,3).index(3)` -> `2` |
| `t1 + t2` | concatenates two tuples | `(1,2) + (3,4)` -> `(1,2,3,4)` |
| `t * n` | repeats the tuple `n` times | `(1,2) * 2` -> `(1,2,1,2)` |
| `x in t` | checks membership | `2 in (1,2,3)` -> `True` |
| `min(t)` / `max(t)` / `sum(t)` | smallest / largest / total | `sum((1,2,3))` -> `6` |


````python
# Example 4: common tuple operations in action
numbers = (4, 8, 15, 16, 23, 42, 8)

print("Length:", len(numbers))
print("Count of 8:", numbers.count(8))
print("Index of 16:", numbers.index(16))
print("Sum:", sum(numbers))
print("Max:", max(numbers))
print("Is 23 in tuple?", 23 in numbers)

combined = numbers + (100, 200)
print("Combined:", combined)
````

### Tuple Unpacking
One of the most useful tuple features: you can assign each item in a tuple to its own variable in a single line.


````python
# Example 5: tuple unpacking
person = ("Alice", 28, "Engineer")

name, age, job = person   # unpacking - each variable gets one value, in order

print(f"{name} is a {age}-year-old {job}.")
````

## 4. Python Dictionaries & Common Dictionary Operations

### What is a dictionary?
A **dictionary** stores data as **key-value pairs**, like a real dictionary maps a word (key) to its meaning (value). Instead of accessing items by position (like lists), you access them by their **key**.

Dictionaries are written with **curly braces `{ }`**:
```python
student = {"name": "Alice", "age": 22, "course": "AI"}
```

Dictionaries are:
- **Unordered by concept** (though since Python 3.7, they preserve insertion order in practice)
- **Mutable** — you can add, update, or remove key-value pairs
- **Keys must be unique** and immutable (strings, numbers, tuples — not lists)


````python
# Example 1: creating and accessing a dictionary
student = {"name": "Alice", "age": 22, "course": "AI"}

print(student)
print(type(student))
print(student["name"])          # access by key -> Alice
print(student.get("age"))       # safer way to access -> 22
print(student.get("grade", "Not Available"))   # default value if key doesn't exist
````

````python
# Example 2: modifying a dictionary
student = {"name": "Alice", "age": 22}

student["age"] = 23              # update an existing key
student["course"] = "AI"         # add a new key-value pair
print(student)

del student["course"]            # remove a key-value pair
print(student)

removed_value = student.pop("age")   # remove and return the value
print("Removed age:", removed_value)
print(student)
````

### Common Dictionary Methods
| Method | What it does | Example |
|---|---|---|
| `.keys()` | returns all keys | `d.keys()` |
| `.values()` | returns all values | `d.values()` |
| `.items()` | returns key-value pairs | `d.items()` |
| `.get(key, default)` | safely fetches a value | `d.get("x", 0)` |
| `.update(d2)` | merges another dict in (overwrites matches) | `d.update({"a":1})` |
| `.pop(key)` | removes key and returns its value | `d.pop("a")` |
| `.clear()` | removes all items | `d.clear()` |
| `key in d` | checks if key exists | `"a" in d` |


````python
# Example 3: common dictionary operations in action
prices = {"apple": 50, "banana": 20, "cherry": 200}

print("Keys:", list(prices.keys()))
print("Values:", list(prices.values()))
print("Items:", list(prices.items()))

prices.update({"date": 150, "apple": 55})   # adds "date", updates "apple"
print("After update:", prices)

print("Does 'banana' exist?", "banana" in prices)
print("Does 'mango' exist?", "mango" in prices)
````

### Nested Dictionaries
Dictionaries can hold other dictionaries (or lists) as values — very common for representing structured, real-world data like a database record.


````python
# Example 4: a nested dictionary
students = {
    "s101": {"name": "Alice", "marks": [85, 90, 78]},
    "s102": {"name": "Bob", "marks": [60, 70, 65]}
}

print(students["s101"]["name"])          # Alice
print(students["s102"]["marks"][0])      # 60 - first mark of Bob

average_alice = sum(students["s101"]["marks"]) / len(students["s101"]["marks"])
print(f"Alice's average: {average_alice:.2f}")
````

````python
val = {"value": 43, 12:111, 10:90}
print(val)
````

## 5. Python Sets & Common Set Operations

### What is a set?
A **set** is an **unordered collection of unique items**. If you add a duplicate value, it's simply ignored. Sets are perfect for removing duplicates and for fast membership testing (`x in my_set`).

Sets are written with **curly braces `{ }`** (like dictionaries, but without key:value pairs):
```python
fruits = {"apple", "banana", "cherry"}
```

⚠️ **Note:** An empty `{}` creates a **dictionary**, not a set. Use `set()` to create an empty set.


````python
# Example 1: creating a set and automatic duplicate removal
numbers = {1, 2, 2, 3, 3, 3, 4}
print(numbers)     # duplicates are automatically removed -> {1, 2, 3, 4}

empty_set = set()   # correct way to create an empty set
empty_dict = {}      # this is a dictionary, NOT a set
print(type(empty_set), type(empty_dict))
````

````python
# Example 2: adding and removing elements
fruits = {"apple", "banana"}

fruits.add("cherry")           # add a single item
print(fruits)

fruits.remove("banana")        # removes item; raises error if not found
print(fruits)

fruits.discard("mango")        # removes item if it exists; NO error if not found
print(fruits)
````

### Mathematical Set Operations
Sets support classic mathematical operations, which makes them very powerful for comparing groups of data.

| Operation | Symbol | Method | Meaning |
|---|---|---|---|
| Union | `\|` | `.union()` | all items from both sets |
| Intersection | `&` | `.intersection()` | items present in **both** sets |
| Difference | `-` | `.difference()` | items in first set but **not** in second |
| Symmetric Difference | `^` | `.symmetric_difference()` | items in either set, but **not both** |


````python
# Example 3: set operations in action
students_python = {"Alice", "Bob", "Charlie", "Diana"}
students_java = {"Charlie", "Diana", "Eve"}

print("Union (all students):", students_python | students_java)
print("Intersection (learn both):", students_python & students_java)
print("Difference (Python only, not Java):", students_python - students_java)
print("Symmetric Difference (only one of the two):", students_python ^ students_java)
````

### Why sets are useful in real projects
- **Removing duplicates** from a list instantly: `list(set(my_list))`
- **Fast membership checks**: checking `x in my_set` is much faster than `x in my_list` for large collections
- **Comparing groups**, like finding common tags, mutual friends, or overlapping skills


````python
# Example 4: removing duplicates from a list using a set
numbers = [1, 3, 2, 3, 5, 1, 4, 2]
unique_numbers = list(set(numbers))
print("Original:", numbers)
print("Unique (order not guaranteed):", unique_numbers)
````

## 6. Iterating over Tuples, Dictionaries & Sets

Just like lists, all three structures can be looped through using `for` loops — but each has its own idioms.


````python
# Example 1: iterating over a tuple (same as a list)
colors = ("red", "green", "blue")

for color in colors:
    print(color)
````

````python
# Example 2: iterating over a dictionary
student = {"name": "Alice", "age": 22, "course": "AI"}

# looping over keys only (default behavior)
for key in student:
    print(key)

print("---")

# looping over key-value pairs (most common and useful pattern)
for key, value in student.items():
    print(f"{key}: {value}")
````

````python
# Example 3: iterating over a set (order is not guaranteed!)
fruits = {"apple", "banana", "cherry"}

for fruit in fruits:
    print(fruit)
````

````python
# Example 4: a practical combined example - counting word frequency using a dictionary
sentence = "the quick brown fox jumps over the lazy dog the fox runs"
words = sentence.split()

word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

for word, count in word_count.items():
    print(f"{word}: {count}")
````

## 7. Choosing the Right Data Structure

| Structure | Ordered? | Mutable? | Duplicates? | Access by | Best for |
|---|---|---|---|---|---|
| **List** `[]` | Yes | Yes | Yes | Index | General-purpose ordered collection you'll modify |
| **Tuple** `()` | Yes | No | Yes | Index | Fixed data that should never change |
| **Dictionary** `{k:v}` | Yes (3.7+) | Yes | Keys must be unique | Key | Mapping labels/IDs to data, fast lookups |
| **Set** `{}` | No | Yes | No (auto-removed) | Value (membership) | Uniqueness checks, math set operations |

### Exercise 1 — Swap Two Variables Without a Temporary Variable (Difficulty: Easy)
**Question:** Swap the values of two variables using **tuple unpacking**, without creating a third "temp" variable.

*Why interviewers ask this:* Tests whether you know Python's tuple-based swapping shortcut — a very common "Python-specific" trick question.


````python
# --- Try it yourself first! ---
a = 5
b = 10
````

**Solution & Explanation:**
- Python evaluates the right-hand side `(b, a)` first, creating a temporary tuple, then unpacks it into `a, b`.
- This works because assignment in Python happens all at once for tuples — there's no risk of overwriting `a` before `b` is read.


````python
a = 5
b = 10

a, b = b, a   # tuple unpacking swap

print("a =", a)
print("b =", b)
````

### Exercise 2 — Find Common Elements Between Two Lists (Difficulty: Easy)
**Question:** Given two lists, find the elements that appear in **both** lists, without using loops (use sets).

*Why interviewers ask this:* Tests knowledge of set operations as an efficient alternative to nested loops.


````python
# --- Try it yourself first! ---
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
````

**Solution & Explanation:**
- Convert both lists to sets, then use the `&` operator (or `.intersection()`) to find common items in one step.
- This is far more efficient than writing a nested loop to compare every pair of elements, especially for large lists.


````python
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

common = set(list1) & set(list2)
print("Common elements:", common)
print("As a list:", list(common))
````

### Exercise 3 — Group Words by Their First Letter (Difficulty: Medium)
**Question:** Given a list of words, build a dictionary where each key is the first letter, and the value is a list of all words starting with that letter.

*Why interviewers ask this:* A classic real-world "grouping" problem that tests dictionaries + the `.setdefault()`/`get()` pattern.


````python
# --- Try it yourself first! ---
words = ["apple", "ant", "banana", "bear", "cat", "car", "dog"]
````

**Solution & Explanation:**
- We loop through each word and grab its first letter with `word[0]`.
- `dict.setdefault(key, [])` returns the existing list for that key, or creates an empty list if the key doesn't exist yet — then we `.append()` the word to it.
- This avoids writing a manual `if key in dict: ... else: ...` check.


````python
words = ["apple", "ant", "banana", "bear", "cat", "car", "dog"]
grouped = {}

for word in words:
    first_letter = word[0]
    grouped.setdefault(first_letter, []).append(word)

for letter, word_list in grouped.items():
    print(f"{letter}: {word_list}")
````

### Exercise 4 — Check if Two Strings are Anagrams (Difficulty: Medium)
**Question:** Given two strings, check if they are **anagrams** of each other (contain the exact same letters with the same frequency, e.g., "listen" and "silent").

*Why interviewers ask this:* A very popular interview question; tests dictionary-based frequency counting.


````python
# --- Try it yourself first! ---
word1 = "listen"
word2 = "silent"
````

**Solution & Explanation:**
- We build a frequency dictionary (letter -> count) for each word.
- If both dictionaries are exactly equal, the words are anagrams.
- We remove spaces and convert to lowercase first, so the check works regardless of case or spacing.


````python
def build_frequency(word):
    freq = {}
    for letter in word.lower().replace(" ", ""):
        freq[letter] = freq.get(letter, 0) + 1
    return freq

word1 = "listen"
word2 = "silent"

is_anagram = build_frequency(word1) == build_frequency(word2)
print(f"'{word1}' and '{word2}' are anagrams: {is_anagram}")
````

### Exercise 5 — Invert a Dictionary (Swap Keys and Values) (Difficulty: Medium)
**Question:** Given a dictionary, create a new dictionary where the keys become values and the values become keys.

*Why interviewers ask this:* Tests dictionary iteration and comprehensions, plus awareness of the "duplicate value" edge case.


````python
# --- Try it yourself first! ---
original = {"a": 1, "b": 2, "c": 3}
````

**Solution & Explanation:**
- We loop through `.items()` to get each key-value pair, and build a new dictionary with `value: key`.
- **Edge case:** if the original dictionary has duplicate values, the inverted dictionary will lose data (later keys overwrite earlier ones), since dictionary keys must be unique. This is worth mentioning in an interview to show awareness of edge cases.


````python
original = {"a": 1, "b": 2, "c": 3}

inverted = {}
for key, value in original.items():
    inverted[value] = key

print(inverted)

# The same logic as a one-line dictionary comprehension
inverted_comprehension = {value: key for key, value in original.items()}
print(inverted_comprehension)
````

### Exercise 6 — Find All Unique Pairs That Sum to a Target (Two Sum, Set Version) (Difficulty: Hard)
**Question:** Given a list of numbers and a target sum, find all unique pairs of numbers that add up to the target. This is a variation of the classic **"Two Sum"** interview question, solved efficiently using a set.

*Why interviewers ask this:* One of the most frequently asked coding interview questions at tech companies; tests whether you can avoid a slow O(n²) nested loop by using a set for O(n) lookups.


````python
# --- Try it yourself first! ---
numbers = [2, 7, 4, 11, 15, -1, 5]
target = 6
````

**Solution & Explanation:**
- We iterate through the list once, keeping a `seen` set of numbers we've already visited.
- For each number, we calculate its `complement` (the value needed to reach the target).
- If the complement is already in `seen`, we've found a pair — we add it as a tuple to `pairs` (using a set of tuples to avoid duplicate pairs).
- We always add the **current number** to `seen` at the end of each loop, so future numbers can find it as a complement.
- This runs in a single pass through the list — much faster than checking every possible pair with nested loops.


````python
numbers = [2, 7, 4, 11, 15, -1, 5]
target = 6

seen = set()
pairs = set()

for num in numbers:
    complement = target - num
    if complement in seen:
        pair = tuple(sorted((num, complement)))   # sorted so (2,4) and (4,2) count as the same pair
        pairs.add(pair)
    seen.add(num)

print("Pairs that sum to", target, ":", pairs)
````
