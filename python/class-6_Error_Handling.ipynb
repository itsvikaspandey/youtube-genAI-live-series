## 1. What is an Error?

An **error** is a problem that prevents our program from behaving as expected.

There are two important categories for beginners:

### Syntax Error
Python cannot understand the code.

Example:

```python
if True
    print("Hello")
```

The program does not even start because the Python syntax is invalid.

### Runtime Error / Exception
The syntax is valid, but something goes wrong while the program is running.

Example:

```python
number = 10
result = number / 0
```

Python understands the code, starts executing it, and then raises `ZeroDivisionError`.

> **Key idea:** Error handling with `try/except` is mainly about handling exceptions that occur during execution.

## 2. Understanding the Traceback

When Python encounters an unhandled exception, it gives us a **traceback**.

A traceback usually tells us:

1. Where the error happened
2. What line caused it
3. What type of exception occurred
4. A message describing the problem

For example:

```text
ZeroDivisionError: division by zero
```

When debugging, don't just look at the last line. Read the traceback from the bottom upward to identify the actual exception and message.

## 4. The `try` and `except` Blocks

The basic pattern is:

```python
try:
    # code that might fail
except:
    # what to do if it fails
```

Think of it like this:

**TRY:** "Run this code."

**EXCEPT:** "If this particular operation fails, handle the problem here."

Let's start with a simple example.


````python
try:
    number = 10
    result = number / 0
    print(result)
except:
    print("Something went wrong!")
````

### Why is this better than allowing the program to crash?

In a real application, an unexpected failure might:

- stop a complete workflow
- terminate a request
- break a data pipeline
- prevent a file from being processed
- give a poor user experience

Error handling lets us decide what should happen when a known problem occurs.


## 5. Catch Specific Exceptions

A **bare `except`** catches almost everything and can hide bugs.

Prefer a specific exception when you know what can go wrong.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
```


````python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
````

````python
try:
    age = int("twenty")
except ValueError:
    print("Please enter a valid number.")
````

## 6. Getting Information From the Exception

We can store the exception object using `as`.

```python
except ValueError as error:
```

This is useful for logging and debugging.


````python
try:
    age = int("twenty")
except ValueError as error:
    print("Error type:", type(error).__name__)
    print("Error message:", error)
````

## 7. Multiple `except` Blocks

Different errors may require different responses.

```python
try:
    ...
except ValueError:
    ...
except ZeroDivisionError:
    ...
```

Python checks the exception handlers from top to bottom.


````python
def divide_numbers(a, b):
    try:
        a = float(a)
        b = float(b)
        return a / b
    except ValueError:
        return "Please provide valid numbers."
    except ZeroDivisionError:
        return "The denominator cannot be zero."

print(divide_numbers("10", "2"))
print(divide_numbers("ten", "2"))
print(divide_numbers("10", "0"))
````

## 8. One `except` for Multiple Exceptions

If the handling logic is identical, multiple exception types can be grouped:

```python
except (ValueError, TypeError):
    ...
```


````python
def convert_to_integer(value):
    try:
        return int(value)
    except (ValueError, TypeError):
        return "Value cannot be converted to an integer."

print(convert_to_integer("25"))
print(convert_to_integer("hello"))
print(convert_to_integer(None))
````

## 9. `else` — Run Code Only When No Exception Occurs

The `else` block runs **only when the `try` block succeeds**.

Structure:

```python
try:
    ...
except SomeError:
    ...
else:
    ...
```

This is useful when you want the success path to be clearly separated from the error path.


````python
def parse_age(value):
    try:
        age = int(value)
    except ValueError:
        print("Invalid age.")
    else:
        print("Age accepted:", age)

parse_age("25")
parse_age("abc")
````

## 10. `finally` — Code That Should Run Anyway

`finally` runs whether an exception happens or not.

It is commonly used for **cleanup**.

Examples:

- closing a file
- releasing a resource
- closing a database connection
- cleaning temporary resources

```python
try:
    ...
except:
    ...
finally:
    ...
```


````python
try:
    print("Opening resource...")
    result = 10 / 2
    print("Result:", result)
except ZeroDivisionError:
    print("Division failed.")
finally:
    print("Cleanup happens here.")
````

````python
try:
    print("Opening resource...")
    result = 10 / 0
    print("Result:", result)
except ZeroDivisionError:
    print("Division failed.")
finally:
    print("Cleanup happens here.")
````

## 11. Complete `try / except / else / finally`

You can combine all four:

```python
try:
    # risky operation
except:
    # error path
else:
    # success path
finally:
    # cleanup
```

The execution flow is:

**try → exception? → except : else → finally**

More precisely:

- If `try` fails → matching `except` runs → `finally`
- If `try` succeeds → `else` runs → `finally`
- `finally` runs in both cases


````python
def process_number(value):
    try:
        number = int(value)
        result = 100 / number
    except ValueError:
        print("Input must be an integer.")
    except ZeroDivisionError:
        print("Number cannot be zero.")
    else:
        print("Calculation successful:", result)
    finally:
        print("Finished processing.\n")

process_number("10")
process_number("abc")
process_number("0")
````

## 12. `raise` — Creating an Exception Yourself

Sometimes Python does not automatically know that a value is invalid for **your business rule**.

For example:

> An age of `-5` is an integer, but it is not a valid age.

We can explicitly raise an exception:

```python
raise ValueError("Age cannot be negative")
```


````python
def validate_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return True

print(validate_age(25))
````

````python
def validate_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return True

try:
    validate_age(-5)
except ValueError as error:
    print("Validation failed:", error)
````

## 13. Custom Exceptions

As applications become larger, custom exceptions can make errors easier to understand.

A custom exception usually inherits from `Exception`.

```python
class InvalidAgeError(Exception):
    pass
```

Then we can raise and catch it like any other exception.


````python
class InvalidAgeError(Exception):
    pass


def validate_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative.")
    if age > 120:
        raise InvalidAgeError("Age looks unrealistic.")
    return True


try:
    validate_age(150)
except InvalidAgeError as error:
    print("Validation error:", error)
````

## 14. Real-World Example — User Input

A common beginner mistake is assuming user input is always valid.

Let's build a safer calculator.


````python
def calculator():
    try:
        first = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        second = float(input("Enter second number: "))

        if operator == "+":
            result = first + second
        elif operator == "-":
            result = first - second
        elif operator == "*":
            result = first * second
        elif operator == "/":
            result = first / second
        else:
            print("Unknown operator.")
            return
        print("user values", first, operator, second)
        print("Result:", result)

    except ValueError:
        print("Please enter valid numbers.")
    except ZeroDivisionError:
        print("Cannot divide by zero.")


# Uncomment to run interactively:
calculator()
````

## 15. Real-World Example — File Handling

File operations can fail for several reasons.

For example:

- file doesn't exist → `FileNotFoundError`
- permission problem → `PermissionError`
- unexpected decoding problem → `UnicodeDecodeError`

We can handle the expected failures explicitly.


````python
def read_file(filename):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            return file.read()
    except FileNotFoundError:
        return f"File not found: {filename}"
    except PermissionError:
        return f"Permission denied: {filename}"

print(read_file("does_not_exist.txt"))
````

## 16. Error Handling in Data / AI Workflows

Error handling becomes especially important in AI applications.

Imagine this pipeline:

**User Input → Document Loading → Chunking → Embedding → Vector DB → LLM → Response**

Any stage can fail.

For example:

```text
Document missing
      ↓
FileNotFoundError
      ↓
Handle error
      ↓
Return useful message
```

Or:

```text
LLM API request
      ↓
Network/API failure
      ↓
Exception
      ↓
Log + retry/fallback
```

The important principle is:

> **Handle expected failures intentionally; don't hide unexpected bugs.**


````python
# Simple AI-style example: document loading

def load_document(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            text = file.read()

        if not text.strip():
            raise ValueError("Document is empty.")

        return text

    except FileNotFoundError:
        print("Document does not exist.")
    except ValueError as error:
        print("Document validation error:", error)
    else:
        print("Document loaded successfully.")
        return text
    finally:
        print("Document loading operation finished.")


load_document("knowledge_base.txt")
````

## 17. Common Mistake — Catching Everything

Avoid this when possible:

```python
try:
    ...
except:
    print("Something went wrong")
```

Why?

Because it can hide programming mistakes.

For example, imagine you accidentally wrote:

```python
result = user_nmae
```

A broad `except` might hide the real `NameError`.

### Better

Catch the errors you actually expect:

```python
except ValueError:
    ...
except FileNotFoundError:
    ...
```

And allow unexpected errors to surface during development.


## 18. Common Mistake — Using Exceptions for Normal Control Flow

Exceptions are not a replacement for ordinary `if/else` logic.

Bad idea:

```python
try:
    if age < 18:
        ...
except:
    ...
```

Use normal conditions for normal business decisions.

Use exceptions when an operation can genuinely fail or when you want to signal an invalid state.

Create a custom exception:

Then create:

```python

```

Rules:

- Amount must be positive.
- Amount cannot be greater than balance.
- Raise `InsufficientBalanceError` when the balance is insufficient.
- Return the new balance when successful.


````python
# TODO: Try it yourself

class InsufficientBalanceError(Exception):
    pass


def withdraw(balance, amount):
    pass
````

### Solution


````python
class InsufficientBalanceError(Exception):
    pass


def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("Withdrawal amount must be positive.")

    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance.")

    return balance - amount


try:
    print("New balance:", withdraw(5000, 1500))
    print("New balance:", withdraw(5000, 6000))
except InsufficientBalanceError as error:
    print("Transaction failed:", error)
except ValueError as error:
    print("Invalid transaction:", error)
````

- `except` → handle an exception

- `finally` → runs regardless of success/failure
- `raise` → explicitly raise an exception
- Custom exceptions → represent application-specific failures
- Catch **specific exceptions** whenever possible
- Don't use exceptions to hide bugs
