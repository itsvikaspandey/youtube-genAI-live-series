## 1. Function Basics (def, parameters, return)

A function is a reusable block of code that performs a specific task. You define it once with `def` and call it as many times as you need.

```python
def greet(name):
    return f"Hello, {name}! Welcome to Blueprint of AI."

message = greet("Vikas")
print(message)
```

```python
# A function with multiple parameters and a computed return value
def add_numbers(a, b):
    total = a + b
    return total

result = add_numbers(5, 7)
print("Sum:", result)

# A function with no return statement returns None by default
def show_message(msg):
    print(msg)

output = show_message("This function prints but doesn't return anything")
print("Return value:", output)
```

## 2. Default, Keyword, *args and **kwargs

These give your functions flexibility — very useful later when calling LLM APIs, which often accept a large number of optional parameters (e.g. `temperature`, `max_tokens`, `model`).

```python
# Default parameter values
def greet_user(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet_user("Vikas"))
print(greet_user("Vikas", greeting="Welcome back"))
```

```python
# Keyword arguments — order doesn't matter when you name the parameter
def create_profile(name, age, city):
    return f"{name}, {age} years old, lives in {city}"

print(create_profile(age=25, city="Moradabad", name="Vikas"))
```

```python
# *args — collects any number of positional arguments into a tuple
def total_marks(*scores):
    print("Scores received:", scores)
    return sum(scores)

print("Total:", total_marks(80, 90, 75, 60))
```

```python
# **kwargs — collects any number of keyword arguments into a dictionary
# This is exactly how you'll call an LLM API — a bunch of named settings
def call_llm_api(prompt, **settings):
    print("Prompt:", prompt)
    print("Settings:", settings)

call_llm_api("Explain Python functions", model="gpt-4", temperature=0.7, max_tokens=500)
```

## 3. Lambda Functions

A lambda is a small, anonymous, single-expression function. Useful for short, throwaway logic — often used inside `sorted()`, `map()`, and `filter()`.

```python
# Regular function vs lambda
def square(x):
    return x * x

square_lambda = lambda x: x * x

print(square(5))
print(square_lambda(5))
```

```python
# Common real use: sorting a list of dictionaries by a specific key
students = [
    {"name": "Aman", "score": 85},
    {"name": "Riya", "score": 92},
    {"name": "Karan", "score": 78},
]

sorted_students = sorted(students, key=lambda s: s["score"], reverse=True)
print(sorted_students)
```

```python
# map() and filter() with lambda
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

squared = list(map(lambda x: x ** 2, numbers))
even_only = list(filter(lambda x: x % 2 == 0, numbers))

print("Squared:", squared)
print("Even only:", even_only)
```

## 4. Scope (Local vs Global Variables)

A variable's scope determines where in your code it can be accessed. Understanding this prevents a very common class of bugs.

```python
# Local scope — variable only exists inside the function
def local_example():
    x = 10  # local variable
    print("Inside function, x =", x)

local_example()
# print(x)  # This would raise a NameError — x doesn't exist outside the function
```

```python
# Global scope — variable defined outside any function, accessible everywhere
counter = 0  # global variable

def increment_wrong():
    counter = counter + 1  # This raises an error — Python treats counter as local here

def increment_correct():
    global counter
    counter += 1

increment_correct()
increment_correct()
print("Counter:", counter)
```

## 5. Recursion

A recursive function calls itself to solve a smaller version of the same problem, until it reaches a base case. Used heavily in tree/graph traversal — which shows up later when working with AI agent decision trees.

```python
# Factorial using recursion
def factorial(n):
    if n == 0 or n == 1:  # base case
        return 1
    return n * factorial(n - 1)  # recursive case

print("5! =", factorial(5))
```

```python
# Fibonacci using recursion
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

fib_sequence = [fibonacci(i) for i in range(10)]
print("First 10 Fibonacci numbers:", fib_sequence)
```

### Exercise 1 — Password Strength Checker (Difficulty: Easy)

Write a function `is_strong_password(password)` that returns `True` if the password is at least 8 characters long AND contains at least one digit. Otherwise return `False`.

```python
# Solution
def is_strong_password(password):
    has_digit = any(char.isdigit() for char in password)
    return len(password) >= 8 and has_digit

print(is_strong_password("abc123"))       # False (too short)
print(is_strong_password("abcdefgh"))     # False (no digit)
print(is_strong_password("abcdefg1"))     # True
```

### Exercise 2 — Flexible Averager using *args (Difficulty: Medium)

Write a function `average(*numbers)` that accepts any number of arguments and returns their average. Handle the case of zero arguments by returning 0.

```python
# Solution
def average(*numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print(average(10, 20, 30))   # 20.0
print(average(5))            # 5.0
print(average())             # 0
```

### Exercise 3 — Build a Mini Config Function using **kwargs (Difficulty: Medium)

Write a function `build_config(**settings)` that returns a dictionary with default values `{"model": "gpt-3.5", "temperature": 0.5}`, but overrides any defaults with values passed in by the caller.

```python
# Solution
def build_config(**settings):
    defaults = {"model": "gpt-3.5", "temperature": 0.5}
    defaults.update(settings)
    return defaults

print(build_config())
print(build_config(temperature=0.9))
print(build_config(model="gpt-4", max_tokens=1000))
```

## Recap

| Concept | Key Idea |
|---|---|
| `def` function | Reusable named block of code |
| Default params | Provide a fallback value if none is passed |
| Keyword args | Pass arguments by name, order-independent |
| `*args` | Collects extra positional args into a tuple |
| `**kwargs` | Collects extra keyword args into a dictionary |
| Lambda | Short, anonymous, single-expression function |
| Local scope | Variable only exists inside its function |
| Global scope | Variable accessible everywhere; needs `global` keyword to modify inside a function |
| Recursion | Function calling itself with a base case |
